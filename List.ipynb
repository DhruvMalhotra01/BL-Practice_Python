The Original List is [29, 45, 32, 49, 37]
The Sum of the list using Custom function is 192 
The Sum of the list using Builtin function is 192 
Comparing the results of Custom function and Builtin function: True


```python
def custom(original_list: list) -> int:
    total = 0
    for num in original_list:
        total += num
    return total


original_list = [29, 45, 32, 49, 37]

print(f"The original list is {original_list}")

custom_sum = custom(original_list)
print(f"The sum of the list using the custom function is {custom_sum}")

builtin_sum = sum(original_list)
print(f"The sum of the list using the built-in function is {builtin_sum}")

print(f"Comparing the results of custom and built-in functions: {custom_sum == builtin_sum}")
```
