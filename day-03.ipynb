PROG 2 : Car Loan Payment

Car_Loan: Write a function CarLoan payment computation. This function takes three arguments - P(Principal Loan Amount), Y(years to pay off) and R (percent interest compounded monthly) and it returns the monthly payment amount.

Purpose : The purpose of this program is to pass the `arguments` to the function

Input => 20000, 5, 5

Output => Monthly car loan payment: 377.42

Hint =>
1. Passing Arguments `monthly_payment = calculate_car_loan_payment(principal_amount, years_to_pay_off, annual_interest_rate)
`
2. Use of formula for calculating the monthly EMI

````python
def monthlyCarLoan(P,R,T):
    monthly_rate = (R / 100) / 12
    
    total_months = T * 12
    amount = P * (monthly_rate * (1 + monthly_rate)**total_months) / ((1 + monthly_rate)**total_months - 1)
    return round(amount, 2)

principal = int(input("Enter loan amount "))
rate = int(input("Enter rate of interest "))
time = int(input("Enter time"))
print(monthlyCarLoan(principal,rate,time))
````

PROG 3: Compute Average

Develop a function to compute average from a given list of numbers and to return the same.

Input => List of numbers `Ex. [10, 20, 30, 40, 50]`

Output => Average of the List is 30.0

Hint =>
- 1: Use `in-built` functions

PROG 4: Return Multiple Values

Develop a function to compute minimum as well  as maximum from  a list of numbers. Function takes number list as input argument and returns both max and min.

Input => num_list = [10, 5, 20, 15, 30]

Output =>
```
Original List : [10, 5, 20, 15, 30]
Minimum number in a List is  5
Maximum number in a List is  30

Hint =>
- 1: Use in-built function min() and max()
- 2: Unpacking the values

````python
def MultipleValues(ls):
    return min(ls), max(ls)

list = {10 ,5 ,20, 15, 30}

min , max = MultipleValues(list)

print(f"Original List : {list}")
print(f"Minimum number in a List is  {min}")
print(f"Maximum number in a List is  {max}")
````

PROG 5: Count Down using Recursion

````python
# PROG 5: Count Down Using Recursion

# Write The Code Here
def countdown(num:  int) -> None:
    print(num)
    if (num == 0):
        return
    countdown(num-1)
    
num = int(input("Enter the number to print the count down: "))
countdown(num)
````

PROG 6: Menu Based School Report [ Percentage / Grade]

````python
# PROG 6: School Report

# Write The Code Here
def calculate_grade(percentage):
    """Return the grade corresponding to a percentage."""
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


def publish_result(
    student_name,
    physics_marks,
    chemistry_marks,
    maths_marks,
    max_marks=100,
    computation_type="p"
):
    """Publish a student's percentage-based or grade-based report."""

    if max_marks <= 0:
        print("Error: Maximum marks must be greater than zero.")
        return

    marks = [physics_marks, chemistry_marks, maths_marks]

    if any(mark < 0 or mark > max_marks for mark in marks):
        print(f"Error: Subject marks must be between 0 and {max_marks}.")
        return

    computation_type = computation_type.lower()

    if computation_type not in ("p", "g"):
        print("Error: Computation type must be 'p' or 'g'.")
        return

    subjects = [
        ("Physics", physics_marks),
        ("Chemistry", chemistry_marks),
        ("Mathematics", maths_marks)
    ]

    print(f"\nNew School Of Learning - Class XI - {student_name}")

    if computation_type == "g":
        print("-" * 34)
        print(f"| {'Subject':^15} | {'Grade':^10} |")
        print("-" * 34)

        for subject, obtained_marks in subjects:
            percentage = obtained_marks / max_marks * 100
            grade = calculate_grade(percentage)
            print(f"| {subject:^15} | {grade:^10} |")

        total_obtained = sum(marks)
        total_maximum = max_marks * 3
        total_percentage = total_obtained / total_maximum * 100
        total_grade = calculate_grade(total_percentage)

        print("-" * 34)
        print(f"| {'Total':^15} | {total_grade:^10} |")
        print("-" * 34)

    else:
        print("-" * 70)
        print(
            f"| {'Subject':^15} | {'Total Marks':^15} | "
            f"{'Marks Obtained':^15} | {'Percentage':^11} |"
        )
        print("-" * 70)

        for subject, obtained_marks in subjects:
            percentage = obtained_marks / max_marks * 100
            percentage_text = f"{percentage:.2f}%"

            print(
                f"| {subject:^15} | {max_marks:^15} | "
                f"{obtained_marks:^15} | {percentage_text:<11} |"
            )

        total_obtained = sum(marks)
        total_maximum = max_marks * 3
        total_percentage = total_obtained / total_maximum * 100
        total_percentage_text = f"{total_percentage:.2f}%"

        print("-" * 70)
        print(
            f"| {'Total':^15} | {total_maximum:^15} | "
            f"{total_obtained:^15} | {total_percentage_text:<11} |"
        )
        print("-" * 70)


# Main program
student_name = input("Enter student's name: ")
max_marks = int(input("Enter max marks per subject: "))
physics_marks = float(input("Enter physics marks: "))
chemistry_marks = float(input("Enter chemistry marks: "))
maths_marks = float(input("Enter maths marks: "))

computation_type = input(
    "Enter computation type (g for grades, p for percentage): "
).strip().lower()

publish_result(
    student_name,
    physics_marks,
    chemistry_marks,
    maths_marks,
    max_marks,
    computation_type
)
````
