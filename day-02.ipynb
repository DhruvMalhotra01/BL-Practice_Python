```python
def add(x: int , y :int ) -> int:
    """" 
    Return the sum of 2 passed parameters
     """
    return x + y

print(add(40, 35))

def add(x,y):
    """
    Return the sum of 2 passed parameters
     """
     
    return x + y

print(add(41, 35))
```

```python
def circle(rad : int) -> float:
    """
    Return the area of circle
    """
    return pi * rad**2

pi = 22/7

rad = int(input("Enter the radius of circle"))
print("Area : ",f"{circle(rad):.4f}")
```

```python
def devide(x: int , y: int) -> int:
    """
    Return the sum of 2 passed parameters
     """
    return x / y

try:
    a = int(input("Enter first number"))
    b = int(input("Enter second number"))
    print("devide : ", devide(a,b))

except ValueError as err:
    print(f"Got error: {err} \n Enter the valid number")
except:
    print("TRY AGAIN")
```
