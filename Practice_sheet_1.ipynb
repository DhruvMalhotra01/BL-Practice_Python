```python
# PROG 1.2: Adding 2 Numbers Measuring Time
import time 

start_time = time.time()

a = 10
b = 4
total = a + b

end_time = time.time()
execution_time = end_time - start_time

print(f"Sum : {total}")
print(f"Execution time: {execution_time} seconds")
# Write the code here
```
