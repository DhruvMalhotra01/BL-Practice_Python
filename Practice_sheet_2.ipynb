Time Conversion

```python
total_seconds = int(input("Enter the time duration in seconds: "))

hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

print(f"\nTime Duration of {total_seconds} seconds in HH:MM:SS format is: {hours:02d}:{minutes:02d}:{seconds:02d}")
```

String Formatting Expression

Tags – {'Item Code':^10} - Center align ‘Item Code’ with length 10

{'-' * 74} - repeat ‘-’ 74 times

{' ' * 54} - repeat space ‘ ’ 54 times

• Unicode Symbol

Tags – universal character encoding, '\u20B9’ – INR ₹

• Python Built-In round()

Tags – round to nearest number
round(10.7) - rounds float to nearest integer 11
round(2.665, 2) - rounds float 2.665 to 2 decimal places, 2.67

• Type Casting

Tags – convert one data type to another, implicit &
Explicit – int(), float(), str()

Prog 6: Bill Generation App Part 1

```python
# Prog 6: Bill Generation App

# Step 1: Setting Constants for Fruits Apple GST as 12% and Orange GST as 5%
APPLE_GST = 0.12      # 12% GST
ORANGE_GST = 0.05     # 5% GST

# Step 2: Enter Buyer Name, Price for Apple and Orange per kg, and quantity
# of Apples and Oranges in kg purchased by the buyer
buyer_name = input("Enter Buyer Name: ")
apple_price_kg = int(input("Enter Apple price per kg: "))
apple_quantity_kg = float(input("Enter Apple Quantity in kg: "))
orange_price_kg = int(input("Enter Orange price per kg: "))
orange_quantity_kg = float(input("Enter Orange Quantity in kg: "))

# Step 3: Calculate Total Price of Apple and Orange
total_price_apple = apple_price_kg * apple_quantity_kg
total_price_orange = orange_price_kg * orange_quantity_kg

# Step 4: Calculate the GST Amount for Apple and Orange
total_gst_apple = total_price_apple * APPLE_GST
total_gst_orange = total_price_orange * ORANGE_GST

# Step 5: Calculate the total billing amount for apple and oranges
# inclusive of GST
total_billing_apple = total_price_apple + total_gst_apple
total_billing_orange = total_price_orange + total_gst_orange

# Step 6: Calculate total amount to be paid by the buyer and round it
total_amount = total_billing_apple + total_billing_orange
total_round_amount = round(total_amount)

# Print the bill
print(f"\nBuyer Name: {buyer_name}")
print(f"{'-' * 74}")
print(
    f"| {'Item Code':^10} | {'Price/Unit':^10} | {'# unit':^5} | "
    f"{'Price':^5} | {'GST':^10} | {'Total w/ GST':^10} |"
)
print(f"{'-' * 74}")

print(
    f"| {'Apple':^10} | {'Rs ' + str(apple_price_kg):^10} | "
    f"{apple_quantity_kg:^5} | "
    f"{'Rs ' + str(total_price_apple):^5} | "
    f"{'Rs ' + format(total_gst_apple, '.2f'):^10} | "
    f"{'Rs ' + str(total_billing_apple):^10} |"
)

print(
    f"| {'Orange':^10} | {'Rs ' + str(orange_price_kg):^10} | "
    f"{orange_quantity_kg:^5} | "
    f"{'Rs ' + str(total_price_orange):^5} | "
    f"{'Rs ' + format(total_gst_orange, '.2f'):^10} | "
    f"{'Rs ' + str(total_billing_orange):^10} |"
)

print(f"{'-' * 74}")
print(f"Total {' ' * 54} \u20B9 {total_amount:.2f}")# rupee symbol \u20B9
print(f"Total Round {' ' * 48} \u20B9 {total_round_amount:.2f}")
print(f"{'-' * 74}")
```

```python
# ============================================
# Student Report Card Management System
# ============================================

# ---------- Step 1: Define Constants ----------
SCHOOL_NAME = "New School Of Learning"
CLASS_NAME = "Class XI"

TOTAL_SUBJECTS = 3
TOTAL_STUDENTS = 3
MAX_MARKS_PER_SUBJECT = 50

subjects = ["Physics", "Chemistry", "Mathematics"]

# These lists will store total marks of all students
subject_totals = [0, 0, 0]
overall_marks = 0


# ---------- Step 2: Take Input for Each Student ----------
for student_number in range(1, TOTAL_STUDENTS + 1):

    print(f"\nEnter details for Student {student_number}")

    student_name = input("Enter Student Name: ")

    marks = []

    # Take marks for each subject
    for subject in subjects:
        mark = int(input(f"Enter {subject} marks (out of {MAX_MARKS_PER_SUBJECT}): "))
        marks.append(mark)

    # ---------- Step 3: Calculate Total and Percentage ----------
    total_marks = sum(marks)

    maximum_marks = TOTAL_SUBJECTS * MAX_MARKS_PER_SUBJECT

    percentage = (total_marks / maximum_marks) * 100

    # Add marks to subject totals
    for i in range(TOTAL_SUBJECTS):
        subject_totals[i] += marks[i]

    overall_marks += total_marks

    # ---------- Step 4: Print Student Report ----------
    print("\n")
    print(f"{SCHOOL_NAME} - {CLASS_NAME} - {student_name}")
    print("-" * 72)

    print(
        f"| {'Subject':^15} | {'Total Marks':^12} | "
        f"{'Marks Obtained':^15} | {'Percentage':^10} |"
    )

    print("-" * 72)

    for i in range(TOTAL_SUBJECTS):

        subject_percentage = (marks[i] / MAX_MARKS_PER_SUBJECT) * 100

        print(
            f"| {subjects[i]:^15} | "
            f"{MAX_MARKS_PER_SUBJECT:^12} | "
            f"{marks[i]:^15} | "
            f"{subject_percentage:^10.2f} |"
        )

    print("-" * 72)

    print(
        f"| {'Total':^15} | "
        f"{maximum_marks:^12} | "
        f"{total_marks:^15} | "
        f"{percentage:^10.2f} |"
    )

    print("-" * 72)


# ---------- Step 5: Calculate Class Average ----------
print("\nClass Average And Percentage For Each Subject:\n")

for i in range(TOTAL_SUBJECTS):

    average = subject_totals[i] / TOTAL_STUDENTS

    average_percentage = (average / MAX_MARKS_PER_SUBJECT) * 100

    print(
        f"{subjects[i]} Average = {average:.2f} "
        f"and Percentage = {average_percentage:.2f}%"
    )


# ---------- Step 6: Overall Class Percentage ----------
overall_percentage = (
    overall_marks /
    (TOTAL_STUDENTS * TOTAL_SUBJECTS * MAX_MARKS_PER_SUBJECT)
) * 100

print(f"\nOverall Class Percentage = {overall_percentage:.2f}%")
```
